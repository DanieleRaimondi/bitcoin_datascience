# **Bitcoin Pi Cycle Top Indicator**

The Pi Cycle Top Indicator compares Bitcoin's 111-day moving average against 2x its 350-day moving average. Whenever the 111-day average crosses above the doubled 350-day average, the market has, in every cycle to date, been within days of the eventual cycle top.

## Data Loading

Bitcoin price is pulled from CoinMetrics (via `functions.fetch_data.fetch_crypto_data`), and both moving averages are computed on top of it.

## Visualization

BTC price (log scale) together with the 111-day SMA and 2x 350-day SMA; every historical crossover is marked with a vertical line.
