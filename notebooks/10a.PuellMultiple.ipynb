# **Bitcoin Puell Multiple**

The Puell Multiple looks at Bitcoin issuance (the USD value of newly mined BTC) relative to its own 365-day moving average. It captures the miner side of the market: when the daily issuance value spikes far above its yearly average, miners are selling into an unusually rich price and cycles have historically topped out shortly after; when it collapses well below average, miner revenue is depressed and cycles have historically bottomed.

## Data Loading

Daily Bitcoin price and total issuance value (USD) are pulled from CoinMetrics (via `functions.fetch_data.fetch_crypto_data`), and the Puell Multiple is derived as `issuance_usd / issuance_usd.rolling(365).mean()`.

## Visualization

A two-panel chart: BTC price (log scale) on top, the Puell Multiple below it, with the classic overheated (>4) and depressed (<0.5) zones shaded in red and green.
