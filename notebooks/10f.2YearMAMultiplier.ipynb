# **Bitcoin 2-Year MA Multiplier**

The 2-Year Moving Average Multiplier plots Bitcoin's price against its own 730-day (2-year) moving average and 5x that average. The 2-year MA has historically acted as a strong long-term accumulation zone, while 5x the 2-year MA has capped the blow-off top of every cycle to date.
