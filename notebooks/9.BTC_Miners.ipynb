# BTC Mining Companies Correlation Analysis

## 🎯 **Goal & Investment Thesis**

This analysis identifies **hedge and diversification opportunities** in the crypto sector by studying correlations between Bitcoin, major mining companies, their Bitcoin holdings, and short interest dynamics.

### **Why Mining Companies are Strategic?**

Mining companies offer **sophisticated Bitcoin exposure** through multiple vectors:

1. **📊 Strong Correlation**: Historical 0.7-1 correlation with BTC
2. **⛏️ Direct Mining**: Active BTC production and accumulation
3. **💼 Balance Sheet Holdings**: Strategic BTC treasury positions
4. **🏗️ Network Infrastructure**: Critical Bitcoin ecosystem players
5. **📈 Operational Leverage**: Amplified exposure to BTC price movements

### **Four-Layer Analysis Framework**

- **Layer 1**: BTC Price Movement (Primary driver)
- **Layer 2**: Stock Price Performance (Market perception + operations)
- **Layer 3**: BTC Holdings (Balance sheet reality)
- **Layer 4**: Short Interest Dynamics (Market sentiment + squeeze potential)

### **Enhanced Strategy with Short Interest**

**Short Interest Analysis Adds**:
- **🎯 Squeeze Opportunities**: High short interest + positive catalysts = potential squeezes
- **📊 Sentiment Gauge**: Short interest as contrarian indicator
- **⏱️ Timing Signals**: Days-to-cover for entry/exit timing
- **🔍 Risk Assessment**: Overleveraged short positions create volatility

### **Trading Strategies**

**Correlation-Based Plays**:
- **📈 Long Miners + Short BTC**: When miners outperform fundamentally
- **📉 Short Miners + Long BTC**: When BTC rises but miners overvalued
- **🔄 Pair Trading**: Exploit temporary divergences

**Short Interest Plays**:
- **🚀 Squeeze Setups**: High short interest + BTC momentum + holdings growth
- **⚖️ Contrarian Positions**: Excessive shorting during accumulation phases
- **📅 Event-Driven**: Target earnings/holdings announcements with high short interest

This approach captures both fundamental Bitcoin exposure and market microstructure dynamics for enhanced alpha generation.

```python
import sys, os
_base = os.getcwd()
while _base and not os.path.isdir(os.path.join(_base, 'functions')):
    _parent = os.path.dirname(_base)
    if _parent == _base: break
    _base = _parent
if os.path.abspath(_base) not in sys.path:
    sys.path.insert(0, os.path.abspath(_base))
from functions.functions_BTC_Miners import generate_plots, update_bitcoin_holdings
import warnings

warnings.filterwarnings("ignore")

ALL_TICKERS = ["MARA", "HUT", "WULF", "RIOT", "BITF", "CLSK", "CANG", "MSTR", "COIN"]


async def complete_btc_analysis(tickers=None, short_interest_source="benzinga"):
    """
    Downloads BTC Holdings data and generates all correlation charts,
    one below the other, in a single call.

    Args:
        tickers (list): Tickers to analyze. If None, uses all (ALL_TICKERS).
        short_interest_source (str): 'benzinga' or 'marketbeat'
    """
    if tickers is None:
        tickers = ALL_TICKERS

    print(f"\n{'='*80}")
    print("🔄 STEP 1: Updating Bitcoin Holdings")
    print(f"{'='*80}")
    print(f"Tickers: {', '.join(tickers)}")
    await update_bitcoin_holdings(tickers=tickers)

    print(f"\n{'='*80}")
    print("📊 STEP 2: Correlation charts")
    print(f"{'='*80}")
    generate_plots(
        tickers=tickers,
        window=365,
        start_date="2017-07-01",
        log_scale=True,
        save_path="../output/BTC_Miners",
        holdings_csv_path="../data/Miners_BTC_Holdings.csv",
        short_interest_source=short_interest_source,
    )
```

```python
await complete_btc_analysis(tickers=[ALL_TICKERS])
```
