# **Bitcoin Address Cohorts Visualizations**

This document describes two functions, `plot_cohorts` and `plot_all_cohorts`, that generate visualizations of Bitcoin address cohorts alongside Bitcoin price movements. Both functions utilize LOESS smoothing to enhance the interpretability of trends in Bitcoin address balances over time.

## 1. `plot_cohorts` Function

The `plot_cohorts` function creates individual charts for various Bitcoin balance cohorts, beginning from the first day with a valid Bitcoin price. Each chart displays the number of addresses within the cohort, their LOESS smoothed curves, and the corresponding Bitcoin price on a logarithmic scale.

### Functionality

- **Data Filtering**: The function filters the input data to start from the first day with a valued Bitcoin price. It also converts the `time` column to a datetime format for accurate plotting.

- **Cohort Definition**: The function defines several relevant cohorts based on the number of addresses with different Bitcoin balances:
    - Addresses with more than 1 million BTC
    - Addresses between 100K and 1M BTC
    - Addresses between 10K and 100K BTC
    - Addresses between 1K and 10K BTC
    - Addresses between 100 and 1K BTC
    - Addresses between 10 and 100 BTC
    - Addresses between 1 and 10 BTC
    - Addresses between 0.1 and 1 BTC
    - Addresses between 0.01 and 0.1 BTC
    - Addresses between 0.001 and 0.01 BTC

- **Plotting**: For each cohort, the number of addresses is plotted alongside the Bitcoin price. LOESS smoothing is applied to the cohort data to visualize trends more clearly. The price is displayed on a logarithmic scale to accommodate Bitcoin's exponential growth.

### Output

The function saves each plot as a JPEG file, providing insights into the trends and behaviors of different Bitcoin address cohorts over time.

---

## 2. `plot_all_cohorts` Function

The `plot_all_cohorts` function generates a single comprehensive chart that combines all relevant Bitcoin balance cohorts, starting from 2012. This visualization includes the number of addresses in each cohort, their LOESS smoothed curves, and the historical Bitcoin price.

### Functionality

- **Data Filtering**: The function filters the input data to include records from January 1, 2012, onward, focusing on valid Bitcoin price entries.

- **Cohort Combination**: 
  - The function aggregates two cohorts for a broader analysis:
    - Addresses with more than 10K BTC.
    - Addresses with balances from 0.001 BTC to 1 BTC.

- **Cohorts Definition**: Several relevant cohorts are defined, similar to the `plot_cohorts` function:
    - More than 10K BTC
    - Between 1K and 10K BTC
    - Between 100 and 1K BTC
    - Between 10 and 100 BTC
    - Between 1 and 10 BTC
    - Between 0.001 and 1 BTC

- **Plotting**: The Bitcoin price is plotted on the primary y-axis using a logarithmic scale. Each cohort's number of addresses is plotted on secondary y-axes, with distinct colors for easy identification. LOESS smoothed curves are included to highlight trends within each cohort.

- **Customization**: The legend is included inside the chart at the bottom right, and layout adjustments ensure that all elements fit neatly within the figure.

### Output

The resulting visualization provides a holistic perspective on the dynamics of Bitcoin address balances across different cohorts in relation to Bitcoin's historical price movements, revealing trends and patterns over time.
