# **Bitcoin Golden Ratio Multiplier**

The Golden Ratio Multiplier tracks Bitcoin's 350-day moving average and its Fibonacci multiples (1.6x, 2x, 3x, 5x, 8x, 13x, 21x). Each multiple band has historically lined up with a resistance level reached during past bull-market advances, giving a rough map of how far a cycle's rally could extend.
