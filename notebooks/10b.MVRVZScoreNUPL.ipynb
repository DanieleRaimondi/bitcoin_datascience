# **Bitcoin MVRV Z-Score and NUPL**

This notebook covers two related on-chain valuation metrics built from the gap between Bitcoin's market cap and its realized cap (the aggregate value of all coins priced at the time they last moved on-chain, a proxy for the network's average cost basis).

## MVRV Z-Score

`(market_cap - realized_cap) / std(market_cap)`. Historically, a Z-Score above ~7 has landed within the euphoric blow-off phase of every cycle top, while a Z-Score below 0 (price trading under the realized/cost-basis price) has marked every cycle bottom.

## NUPL (Net Unrealized Profit/Loss)

`(market_cap - realized_cap) / market_cap`. This is the same gap expressed as a fraction of market cap, split into the five classic sentiment zones: Capitulation, Hope/Fear, Optimism, Belief and Euphoria/Greed.
