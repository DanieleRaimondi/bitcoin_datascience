# **Bitcoin Hash Ribbons**

The Hash Ribbons indicator (Charles Edwards) compares the 30-day and 60-day moving averages of Bitcoin's hashrate. When the 30-day average drops below the 60-day average, unprofitable miners are capitulating and shutting down; when it recovers back above, the capitulation is over. Historically, the recovery point has marked strong accumulation zones.

## Data Loading

Bitcoin price and hashrate are pulled from CoinMetrics (via `functions.fetch_data.fetch_crypto_data`).

## Visualization

Price with capitulation periods shaded in red and each recovery signal marked with a vertical line, plus the two hashrate moving averages below it.
